# notebook 4 — predicting future food prices using past prices

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('wfp_food_prices_clean.csv')
df['date'] = pd.to_datetime(df['date'])
print('loaded:', df.shape)
```

```python
# focus on maize (white) — the most common commodity across all regions
# compute the national monthly average price
maize = df[df['commodity'] == 'Maize (white)']
monthly = maize.groupby('date')['price'].mean().reset_index()
monthly = monthly.sort_values('date').reset_index(drop=True)
monthly.columns = ['date', 'price']

# plot the full time series
monthly.set_index('date')['price'].plot(color='#1B4D3E', linewidth=1.8, figsize=(12,4))
plt.title('Maize (white) — monthly average price across Cameroon')
plt.ylabel('Price (XAF)'); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
print('months available:', len(monthly))
```

```python
# create lag features: give the model the last 6 months as inputs
for lag in range(1, 7):
    monthly[f'lag_{lag}'] = monthly['price'].shift(lag)
monthly['year']  = monthly['date'].dt.year
monthly['month'] = monthly['date'].dt.month
monthly = monthly.dropna().reset_index(drop=True)

feature_cols = [f'lag_{i}' for i in range(1,7)] + ['year', 'month']
X = monthly[feature_cols]
y = monthly['price']
print('features used:', feature_cols)
print('total samples:', len(X))
```

```python
# temporal split — train on old data, test on recent data
split = int(len(X) * 0.80)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]
dates_test = monthly['date'].iloc[split:]

print(f'training on {len(X_train)} months')
print(f'testing on  {len(X_test)} months ({dates_test.min().date()} to {dates_test.max().date()})')
```

```python
# train the model
rf = RandomForestRegressor(n_estimators=200, max_depth=10, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae  = mean_absolute_error(y_test, y_pred)
r2   = r2_score(y_test, y_pred)

print(f'RMSE : {rmse:.1f} XAF  (average prediction error)')
print(f'MAE  : {mae:.1f} XAF  (typical absolute error)')
print(f'R²   : {r2:.2f}       (how much variance is explained)')
```

```python
# plot actual vs predicted
plt.figure(figsize=(12,4))
plt.plot(dates_test.values, y_test.values, label='Actual', color='#1B4D3E', linewidth=2)
plt.plot(dates_test.values, y_pred, label='Predicted', color='#E69F00', linestyle='--', linewidth=2)
plt.title('Maize price — actual vs predicted on test period')
plt.xlabel('Date'); plt.ylabel('Price (XAF)')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

```python
# feature importance — which lag matters most?
imp = pd.Series(rf.feature_importances_, index=feature_cols).sort_values(ascending=True)
imp.plot(kind='barh', color='#1B4D3E', edgecolor='white', figsize=(7,4))
plt.title('Feature importance — lag_1 (last month) is the strongest signal')
plt.xlabel('Importance'); plt.tight_layout(); plt.show()
```
